```python

class employee:
    def __init__(self):
        self.employeeid = int(input("Enter your id: "))
        self.employeename = input("Enter your name: ")

class salary(employee):
    def __init__(self):
        super().__init__()
        self.bonus = 1000
        self.salary = float(input("Enter your salary: "))

    def total(self):
        totalsal = self.salary + self.bonus

        print(f"Employee name is: {self.employeename}")
        print(f"Employee ID is: {self.employeeid}")
        print(f"Total salary is: {totalsal}")

empob = salary()
empob.total()
```

```python
class vehicle:
  def __init__(self):
    self.name=input("Enter your vehicle: ")
    self.brand=input("Enter your brand: ")
    self.model=input("Enter your model: ")

class car(vehicle):
  def __init__(self):
    super().__init__()
    self.fueltype=input("Enter your fuel type: ")
    self.seatingcapacity=input("Enter your seating capacity: ")
  def vehiclede(self):
    print(f" vehicle name is{self.name}\n Brand:{self.brand}\n model:{self.model}")
    print(f" fuel type{self.fueltype}\n seating capacity{self.seatingcapacity}")

object=car()
object.vehiclede()


```
