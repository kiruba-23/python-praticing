```python
Num=set()
for i in range(1,11):
  num=int(input("Enter your number:"))
  Num.add(num)
print(Num)
Num.add(0)
print(Num)
Num.discard(90)
print(Num)
print(1 in Num)
Num1={2,4,6,8,10}
Numuni=Num.union(Num1)
print(Numuni)
Numin=Num.intersection(Num1)
print(Numin)
Numdiff=Num.difference(Num1)
print(Numdiff)
Numsy=Num.symmetric_difference(Num1)
print(len(Numsy))
```
